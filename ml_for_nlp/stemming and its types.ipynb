# Stemming and its type in the NLP

```python
word = ["goes", "gone", "going", "eating", "eats", "eaten", "history", "historical", "programs", "programming"]
```

### PorterStemmer

```python
from nltk.stem import PorterStemmer
```

```python
stemming = PorterStemmer()
```

```python
for w in word:
    print(w+"--->"+ stemming.stem(w))
```

```python
stemming.stem("Congrtulations")
```

### RegexpStemmer class

```python
from nltk.stem import RegexpStemmer
```

```python
reg_stemmer = RegexpStemmer('ing$|s$|e$|able$|es$|en$', min=4)
```

```python
for i in word:
    print(i+"--->"+reg_stemmer.stem(i))
    
```

```python
reg_stemmer.stem("eating")
```

### Snowball Stemmer

```python
from nltk.stem import SnowballStemmer
snowballstemmer = SnowballStemmer('english')
```

```python
for k in word:
    print(k+"--->"+snowballstemmer.stem(k))
    
```

```python
snowballstemmer.stem("fairly")
```
