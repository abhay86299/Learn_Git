```python
word = """Hey! this is nltk the very famous library of the NLP in AI, This is 'used' to create the token's of the words.
This can't be used without importing the values. They'll make it better as the time passes.
"""
```

```python
from nltk import sent_tokenize, word_tokenize 
```

```python
print(sent_tokenize(word))
print(word_tokenize(word))
```

```python
sentences = sent_tokenize(word)
for i in sentences:
    print(i)
```

```python
word_s = word_tokenize(word)
for i in word_s:
    if i == ".":
        continue
    else:
        print(i)
        
```

```python
from nltk.tokenize import TreebankWordTokenizer 
t = TreebankWordTokenizer()
t.tokenize(word)
```

```python
from nltk.tokenize import PunktSentenceTokenizer
tokenizer = PunktSentenceTokenizer()
s = tokenizer.tokenize(word)
```

```python
print(s)
```
