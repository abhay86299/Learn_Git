```python
import nltk
nltk.download('wordnet')
```

```python
from nltk.stem import WordNetLemmatizer
```

```python
lemmatizer = WordNetLemmatizer()
```

```python
"""
Valid options are 
`"n"` for nouns,
`"v"` for verbs, 
`"a"` for adjectives, 
`"r"` for adverbs, 
`"s"` for satellite adjectives
"""
lemmatizer.lemmatize("going", pos="v")
```

```python
words = ["goes", "gone", "going", "eating", "eats", "eaten", "history", "historical", "programs", "programming"]
```

```python
for word in words:
    print(word+"--->"+lemmatizer.lemmatize(word, pos="v"))
```
